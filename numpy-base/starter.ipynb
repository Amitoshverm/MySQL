list vs numpy array

```python
py_list = ([10, 20, 30, 40, 50])
print("python multiplication ", py_list * 2)
```

```python
import time 
start = time.time()
py_list = (i * 2 for i in range(1000000))
print("\n list operation time: ", time.time() - start)


start = time.time()
np_arr = np.arange(1000000) * 2
print("\n numpy operation time: ", time.time() - start)
```

```python
zeros = np.zeros((3, 4))
print("zeros: \n", zeros)

ones = np.ones((2, 3))
print("one: \n", ones)

full = np.full((2, 3), 5)
print("full: \n", full)

random = np.random.random((3, 3))
print("random: \n", random)

sequence = np.arange(0, 10, 4)
print("sequence: \n", sequence)


```
